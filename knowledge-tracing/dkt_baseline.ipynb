## Comparison with Published Results

My DKT implementation achieved a best test AUC of **0.8201** on
ASSISTments 2009.

This result is very close to the **0.8215 ± 0.0005** AUC reported by
HCNOH for DKT on ASSISTments 2009. However, the results are not
directly identical experiments because the HCNOH repository uses
different training configurations, including a 90/10 train-test
ratio, whereas this experiment uses an 80/20 student-level split.

The original DKT paper by Piech et al. reported an AUC of **0.86**
on ASSISTments. The difference between this result and our result
likely reflects differences in data preprocessing, train-test
splitting, model configuration, and training procedure. Therefore,
the published AUC is used as a reference point rather than as a
directly controlled benchmark.

Overall, the close agreement between our result and the HCNOH
implementation provides evidence that our DKT baseline is
implemented and trained correctly.
