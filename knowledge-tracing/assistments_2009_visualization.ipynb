# Visualization of ASSISTments 2009 data

```python
## Imports
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import drive
drive.mount('/content/drive')

path = "/content/drive/MyDrive/education-ml-research/ASSISTments2009/skill_builder_data.csv"
```

## Load and verify dataset

```python
df = pd.read_csv(path, encoding='latin1')

print(df.head())

print(df.info())

print(df.shape)
```

## Visualize data

```python
## Attempts per student distribution

attempts_per_student = df.groupby("user_id").size()

plt.figure(figsize=(8,5))
plt.hist(attempts_per_student, bins=50)
plt.title("Distribution of Attempts per Student")
plt.xlabel("Number of Attempts")
plt.ylabel("Number of Students")

plt.show()
```

```python
## Correct vs incorrect responses by skill for 20 most common skills

# Find the 20 most common skills
top_skills = df["skill_id"].value_counts().head(20).index

# Keep only those skills
top_df = df[df["skill_id"].isin(top_skills)]

# Count correct and incorrect responses for each skill
skill_correct = (
    top_df.groupby("skill_name")["correct"]
          .value_counts()
          .unstack(fill_value=0)
)

# Rename the columns for readability
skill_correct.columns = ["Incorrect", "Correct"]

# Plot
plt.figure(figsize=(14,14))

skill_correct.plot(
    kind="bar",
    stacked=True,
    figsize=(14,6)
)

plt.title("Correct vs Incorrect Responses for Top 20 Most Common Skills")
plt.xlabel("Skill ID")
plt.ylabel("Number of Attempts")

plt.legend(title="Response")

plt.tight_layout()

plt.show()
```

```python
## Average accuracy vs number of attempts per student

# Number of attempts per student
attempts = df.groupby("user_id").size()

# Average accuracy per student
accuracy = df.groupby("user_id")["correct"].mean()

# Combine into one DataFrame
student_stats = pd.DataFrame({
    "Attempts": attempts,
    "Accuracy": accuracy
})

# Plot
plt.figure(figsize=(15,6))

plt.scatter(
    student_stats["Attempts"],
    student_stats["Accuracy"],
    alpha=0.5
)

plt.title("Student Accuracy vs Number of Attempts")
plt.xlabel("Number of Attempts")
plt.ylabel("Average Accuracy")

plt.grid(True)

plt.tight_layout()

plt.show()
```

```python
## 20 most common skills

top_skills = (
    df["skill_name"]
    .value_counts()
    .head(20)
)

plt.figure(figsize=(12,8))

plt.bar(
    top_skills.index.astype(str),
    top_skills.values
)

plt.xticks(rotation=90)

plt.title("Top 20 Most Common Skills")
plt.xlabel("Skill")
plt.ylabel("Attempts")

plt.tight_layout()

plt.show()
```

```python
## Student accuracy distribution

student_accuracy = (
    df.groupby("user_id")["correct"]
      .mean()
)

plt.figure(figsize=(8,5))

plt.hist(student_accuracy, bins=30)

plt.title("Distribution of Student Accuracy")
plt.xlabel("Accuracy")
plt.ylabel("Students")

plt.show()
```

```python
## Student accuracy by ability quartiles

## Compute student accuracy
student_accuracy = (
    df.groupby("user_id")["correct"]
      .mean()
)

student_df = student_accuracy.reset_index()

student_df.columns = ["user_id", "accuracy"]

## Assign quartiles
student_df["quartile"] = pd.qcut(
    student_df["accuracy"],
    q=4,
    labels=[
        "Q1 (Lowest)",
        "Q2",
        "Q3",
        "Q4 (Highest)"
    ]
)

## Plot
plt.figure(figsize=(8,6))

student_df.boxplot(
    column="accuracy",
    by="quartile",
    grid=False
)

plt.title("Student Accuracy by Ability Quartile")
plt.suptitle("")

plt.xlabel("Ability Quartile")
plt.ylabel("Average Accuracy")

plt.tight_layout()

plt.show()
```
