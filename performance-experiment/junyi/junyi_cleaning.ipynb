# Cleaning the Junyi Academy dataset

```python
# Import the dataset and view the structure

import pandas as pd
import numpy as np

FILE_PATH = "junyi_raw.csv"

df = pd.read_csv(FILE_PATH, low_memory=False)

# Dataset dimensions
print("Shape:", df.shape)

# Log raw dataset
raw_rows = len(df)
raw_students = df["uuid"].nunique()

print("=== Junyi: Raw Dataset ===")
print(f"Raw rows: {raw_rows:,}")
print(f"Raw students: {raw_students:,}")
print(f"Raw columns: {len(df.columns)}")

# Column names
print("\nColumns:")
print(df.columns.tolist())

# First 10 rows
display(df.head(10))
```

```python
# Add a in_subsample column to the dataset to indicate whether a student is in the subsample or not
# Subsample is created due to computational constraints. The subsample is created by randomly selecting 25,000 students from the dataset.
SEED = 42
N_SUBSAMPLE = 25_000

# Get unique students
unique_students = df["uuid"].dropna().unique()

assert len(unique_students) >= N_SUBSAMPLE

# Randomly select 25,000 students
rng = np.random.default_rng(SEED)

sampled_students = rng.choice(
    unique_students,
    size=N_SUBSAMPLE,
    replace=False
)

# Add subsample indicator
df["in_subsample"] = (
    df["uuid"].isin(sampled_students).astype(int)
)

# Verify
assert df.loc[df["in_subsample"] == 1, "uuid"].nunique() == 25_000
assert df.groupby("uuid")["in_subsample"].nunique().max() == 1

print("=== Junyi Subsample Summary ===")
print(f"Total students: {len(unique_students):,}")
print(f"Subsample students: {N_SUBSAMPLE:,}")
print(f"Subsample interactions: {(df['in_subsample'] == 1).sum():,}")
print(f"Remaining interactions: {(df['in_subsample'] == 0).sum():,}")

display(df.head())
```

```python
# 1. Record initial counts
initial_rows = len(df)
initial_students = df["uuid"].nunique()

print("=== Junyi Raw Dataset ===")
print(f"Rows: {initial_rows:,}")
print(f"Students: {initial_students:,}")

# 2. Rename columns
df = df.rename(columns={
    "uuid": "user_id",
    "upid": "question_id",
    "ucid": "skill_id",
    "timestamp_TW": "timestamp",
    "is_correct": "correct"
})

# 3. Keep essential columns
columns_to_keep = [
    "user_id",
    "question_id",
    "skill_id",
    "timestamp",
    "problem_number",
    "exercise_problem_repeat_session",
    "total_attempt_cnt",
    "used_hint_cnt",
    "correct",
    "in_subsample"
]

df = df[columns_to_keep].copy()

# 4. Convert timestamps to UTC Unix seconds
timestamps = pd.to_datetime(
    df["timestamp"],
    errors="coerce",
    utc=True
)

# Store Unix seconds; invalid timestamps become missing
df["timestamp"] = (
    timestamps.astype("int64")
    .where(timestamps.notna(), pd.NA)
    .floordiv(10**9)
    .astype("Int64")
)

# 5. Remove rows missing essential values
required_cols = [
    "user_id",
    "question_id",
    "skill_id",
    "timestamp",
    "correct"
]

missing_mask = df[required_cols].isna().any(axis=1)
missing_removed = int(missing_mask.sum())

df = df.loc[~missing_mask].copy()

# 6. Remove invalid correctness values
df["correct"] = pd.to_numeric(
    df["correct"],
    errors="coerce"
)

valid_correct = df["correct"].isin([0, 1])
invalid_correct_removed = int((~valid_correct).sum())

df = df.loc[valid_correct].copy()
df["correct"] = df["correct"].astype(int)

# 7. Remove noskill rows
noskill_mask = (
    df["skill_id"]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("noskill")
    .fillna(False)
)

noskill_removed = int(noskill_mask.sum())
df = df.loc[~noskill_mask].copy()

# 8. Reset index
df = df.reset_index(drop=True)

# 9. Verification
assert df[required_cols].notna().all().all()
assert df["correct"].isin([0, 1]).all()
assert df["timestamp"].dtype == "Int64"
assert df["in_subsample"].isin([0, 1]).all()

print("\n=== Junyi Cleaning Summary ===")
print(f"Initial rows: {initial_rows:,}")
print(f"Missing required values removed: {missing_removed:,}")
print(f"Invalid correctness removed: {invalid_correct_removed:,}")
print(f"Noskill rows removed: {noskill_removed:,}")
print(f"Final rows: {len(df):,}")
print(f"Total rows removed: {initial_rows - len(df):,}")

print(f"\nInitial students: {initial_students:,}")
print(f"Remaining students: {df['user_id'].nunique():,}")
print(f"Subsample students remaining: {df.loc[df['in_subsample'] == 1, 'user_id'].nunique():,}")

display(df.head(10))
```

```python
# Record counts before deduplication
rows_before = len(df)
students_before = df["user_id"].nunique()

# Count exact duplicates
duplicate_count = df.duplicated().sum()

# Remove exact duplicates
df = df.drop_duplicates().reset_index(drop=True)

# Cleaning summary
print("=== Junyi Duplicate Removal ===")
print(f"Rows before: {rows_before:,}")
print(f"Exact duplicates removed: {duplicate_count:,}")
print(f"Rows remaining: {len(df):,}")
print(f"Students before: {students_before:,}")
print(f"Students remaining: {df['user_id'].nunique():,}")

# Verify
assert not df.duplicated().any()

print("\nAll exact duplicate rows removed!")
```

```python
# Preserve original row order as a final deterministic tie-breaker
df["_original_row"] = np.arange(len(df))

# Timestamps are only accurate to the nearest 15 minutes, so we sort by user_id, timestamp, 
# problem_number, exercise_problem_repeat_session, and original row order to ensure a 
# consistent chronological order for each student.
df = (
    df.sort_values(
        [
            "user_id",
            "timestamp",
            "problem_number",
            "exercise_problem_repeat_session",
            "_original_row"
        ],
        kind="mergesort",
        na_position="last"
    )
    .drop(columns="_original_row")
    .reset_index(drop=True)
)

print("Junyi chronological sorting complete!")
print(f"Students: {df['user_id'].nunique():,}")
print(f"Interactions: {len(df):,}")

display(
    df[
        [
            "user_id",
            "timestamp",
            "skill_id",
            "question_id",
            "problem_number",
            "exercise_problem_repeat_session"
        ]
    ].head(20)
)
```

```python
window_counts = (
    df.groupby(["user_id", "timestamp"])
    .size()
)

ambiguous_windows = window_counts[window_counts > 1]

print("=== Timestamp Window Analysis ===")
print(f"Total student-time windows: {len(window_counts):,}")
print(f"Windows with multiple interactions: {len(ambiguous_windows):,}")
print(f"Interactions in these windows: {ambiguous_windows.sum():,}")
print(f"Percentage of interactions affected: {ambiguous_windows.sum() / len(df) * 100:.2f}%")
```

```python
ordering_cols = [
    "user_id",
    "timestamp",
    "problem_number",
    "exercise_problem_repeat_session"
]

tie_counts = df.groupby(
    ordering_cols,
    dropna=False
).size()

print("\n=== Ordering Tie Analysis ===")
print(f"Groups with identical ordering keys: {(tie_counts > 1).sum():,}")
print(f"Rows in tied groups: {tie_counts[tie_counts > 1].sum():,}")
```

```python
# Count distinct exercises per student-timestamp window
exercise_counts = (
    df.groupby(["user_id", "timestamp"])["skill_id"]
    .nunique()
)

multi_exercise_windows = exercise_counts[
    exercise_counts > 1
]

window_sizes = (
    df.groupby(["user_id", "timestamp"])
    .size()
)

affected_rows = window_sizes.loc[
    multi_exercise_windows.index
].sum()

print("=== Cross-Exercise Ordering Ambiguity ===")
print(f"Total windows: {len(exercise_counts):,}")
print(f"Windows with multiple exercises: {len(multi_exercise_windows):,}")
print(f"Rows in multi-exercise windows: {affected_rows:,}")
print(f"Percentage of all rows: {affected_rows / len(df) * 100:.2f}%")
```

```python
# The interactions are able to be sorted chronologically within each skill, but not throughout the entire time group.
df = (
    df.sort_values(
        [
            "user_id",
            "timestamp",
            "skill_id",
            "problem_number",
            "exercise_problem_repeat_session"
        ],
        kind="mergesort",
        na_position="last"
    )
    .reset_index(drop=True)
)

# Verify timestamps are chronological within students
assert df.groupby("user_id")["timestamp"].apply(
    lambda x: x.is_monotonic_increasing
).all()

print("Junyi approximate chronological ordering complete!")
print(f"Students: {df['user_id'].nunique():,}")
print(f"Interactions: {len(df):,}")
```

```python
# A continuity-aware ordering is being ran to compare the skills within neighboring time groups to establish a better ordering.

# ------------------------------------------------------------
# 1. Construct exercise blocks
# ------------------------------------------------------------

blocks = []

for (user_id, timestamp, skill_id), group in df.groupby(
    ["user_id", "timestamp", "skill_id"],
    sort=False
):
    blocks.append({
        "user_id": user_id,
        "timestamp": timestamp,
        "skill_id": skill_id,
        "row_indices": group.index.to_numpy()
    })

blocks_df = pd.DataFrame(blocks)

# ------------------------------------------------------------
# 2. Continuity-aware ordering
# ------------------------------------------------------------

ordered_indices = []

for user_id, student_blocks in blocks_df.groupby(
    "user_id", sort=False
):

    windows = [
        (timestamp, window.copy())
        for timestamp, window in student_blocks.groupby(
            "timestamp", sort=True
        )
    ]

    previous_last_skill = None

    for window_idx, (timestamp, window) in enumerate(windows):

        available_skills = window["skill_id"].tolist()
        ordered_skills = []

        # A. Continue the previous exercise if possible
        if previous_last_skill in available_skills:
            ordered_skills.append(previous_last_skill)
            available_skills.remove(previous_last_skill)

        # B. Identify exercises appearing in the next window
        next_skills = set()

        if window_idx + 1 < len(windows):
            next_skills = set(
                windows[window_idx + 1][1]["skill_id"]
            )

        # C. Prefer a continuing exercise at the end
        next_candidates = [
            skill for skill in available_skills
            if skill in next_skills
        ]

        if next_candidates:
            last_skill = sorted(
                next_candidates,
                key=str
            )[0]
            available_skills.remove(last_skill)
        else:
            last_skill = None

        # D. Order remaining exercises deterministically
        ordered_skills.extend(
            sorted(available_skills, key=str)
        )

        if last_skill is not None:
            ordered_skills.append(last_skill)

        # E. Append interactions in reconstructed order
        block_lookup = {
            row["skill_id"]: row["row_indices"]
            for _, row in window.iterrows()
        }

        for skill in ordered_skills:
            ordered_indices.extend(
                block_lookup[skill].tolist()
            )

        previous_last_skill = ordered_skills[-1]

# ------------------------------------------------------------
# 3. Apply reconstructed ordering
# ------------------------------------------------------------

assert len(ordered_indices) == len(df)
assert len(set(ordered_indices)) == len(df)

df = df.iloc[ordered_indices].reset_index(drop=True)

# ------------------------------------------------------------
# 4. Verification
# ------------------------------------------------------------

# Students remain grouped
assert df["user_id"].is_monotonic_increasing if pd.api.types.is_numeric_dtype(
    df["user_id"]
) else True

# Timestamps remain chronological within students
assert df.groupby("user_id")["timestamp"].apply(
    lambda x: x.is_monotonic_increasing
).all()

# Problem numbers increase within each student/exercise/window
assert df.groupby(
    ["user_id", "timestamp", "skill_id"]
)["problem_number"].apply(
    lambda x: x.is_monotonic_increasing
).all()

print("=== Junyi Continuity-Aware Ordering Complete ===")
print(f"Students: {df['user_id'].nunique():,}")
print(f"Interactions: {len(df):,}")
print(f"Exercise blocks: {len(blocks_df):,}")
print("All ordering checks passed!")

display(
    df[
        [
            "user_id",
            "timestamp",
            "skill_id",
            "question_id",
            "problem_number",
            "exercise_problem_repeat_session"
        ]
    ].head(20)
)
```

```python
# ============================================================
# INSPECT SKILL TRANSITIONS IN RECONSTRUCTED JUNYI DATA
# ============================================================

# Identify rows where the skill changes within the same student
skill_changed = df["skill_id"].ne(df["skill_id"].shift())

same_student = df["user_id"].eq(df["user_id"].shift())

# Find transitions
transition_indices = np.flatnonzero(
    (skill_changed & same_student).to_numpy()
)

print(f"Total skill transitions: {len(transition_indices):,}")

# Find a transition where both skills share the same timestamp
same_timestamp = df["timestamp"].eq(df["timestamp"].shift())

within_window_indices = np.flatnonzero(
    (skill_changed & same_student & same_timestamp).to_numpy()
)

print(f"Transitions within the same timestamp: {len(within_window_indices):,}")

# Select the first within-window transition
assert len(within_window_indices) > 0, "No within-window transitions found."

idx = within_window_indices[0]

# Show 5 rows before and 5 rows after
columns = [
    "user_id",
    "timestamp",
    "skill_id",
    "question_id",
    "problem_number",
    "exercise_problem_repeat_session"
]

print("\n=== SKILL TRANSITION EXAMPLE ===")
print(f"Student: {df.loc[idx, 'user_id']}")
print(f"Previous skill: {df.loc[idx - 1, 'skill_id']}")
print(f"New skill: {df.loc[idx, 'skill_id']}")

display(
    df.iloc[max(0, idx - 5):idx + 6][columns]
)
```

```python
# Calculates each student's early window as the larger of
# 40% of interactions or the number needed to reach 20 unique questions.
# Uses the existing chronological row order.

def early_window_lengths(
    df,
    student_col="user_id",
    question_col="question_id",
    early_fraction=0.40,
    min_unique_questions=20
):
    results = []

    for student_id, group in df.groupby(student_col, sort=False):

        # Use existing row order (already chronologically sorted)
        n_rows = len(group)
        n_unique_questions = group[question_col].nunique()

        # Minimum number of rows covering 40% of interactions
        fraction_length = int(np.ceil(early_fraction * n_rows))

        # Find the row where the 20th distinct question appears
        first_encounters = ~group[question_col].duplicated()

        if n_unique_questions >= min_unique_questions:
            twentieth_index = np.flatnonzero(
                first_encounters.to_numpy()
            )[min_unique_questions - 1]

            rows_to_twentieth = int(twentieth_index + 1)

            early_length = max(
                fraction_length,
                rows_to_twentieth
            )
        else:
            rows_to_twentieth = pd.NA
            early_length = pd.NA

        remaining_rows = (
            n_rows - early_length
            if pd.notna(early_length)
            else pd.NA
        )

        results.append({
            student_col: student_id,
            "total_rows": n_rows,
            "unique_questions": n_unique_questions,
            "fraction_length": fraction_length,
            "rows_to_20th_question": rows_to_twentieth,
            "early_window_length": early_length,
            "remaining_rows": remaining_rows
        })

    return pd.DataFrame(results)
```

```python
# Calculate early-window lengths
early_windows = early_window_lengths(df)

# Assign exclusion reasons
early_windows["exclusion_reason"] = np.select(
    [
        early_windows["unique_questions"] < 20,
        early_windows["remaining_rows"].fillna(-1) < 5
    ],
    [
        "fewer than 20 different questions",
        "fewer than 5 rows left"
    ],
    default=""
)

early_windows["eligible"] = (
    early_windows["exclusion_reason"] == ""
)

# Keep eligible student summaries
kept_students_df = early_windows[
    early_windows["eligible"]
].copy()

# Keep excluded student summaries and reasons
excluded_students_df = early_windows[
    ~early_windows["eligible"]
].copy()

# Filter the canonical interaction dataset
df_filtered = df[
    df["user_id"].isin(
        kept_students_df["user_id"]
    )
].copy().reset_index(drop=True)

print("Students before:", df["user_id"].nunique())
print("Students retained:", len(kept_students_df))
print("Students excluded:", len(excluded_students_df))

print("\nRows before:", len(df))
print("Rows retained:", len(df_filtered))
print("Rows excluded:", len(df) - len(df_filtered))

print("\nExclusion reasons:")
print(excluded_students_df["exclusion_reason"].value_counts())

display(excluded_students_df.head(10))
```

```python
# 1. Create filtered subsample
df_filtered_subsample = (
    df_filtered[df_filtered["in_subsample"] == 1]
    .copy()
    .reset_index(drop=True)
)

# 2. Save both datasets
df_filtered.to_parquet(
    "junyi_filtered.parquet",
    index=False
)

df_filtered_subsample.to_parquet(
    "junyi_filtered_subsample.parquet",
    index=False
)

# 3. Reload and verify
full_check = pd.read_parquet(
    "junyi_filtered.parquet"
)

subsample_check = pd.read_parquet(
    "junyi_filtered_subsample.parquet"
)

pd.testing.assert_frame_equal(
    df_filtered,
    full_check,
    check_dtype=False
)

pd.testing.assert_frame_equal(
    df_filtered_subsample,
    subsample_check,
    check_dtype=False
)

assert (subsample_check["in_subsample"] == 1).all()

assert set(subsample_check["user_id"]).issubset(
    set(full_check["user_id"])
)

# 4. Summary
print("=== Junyi Datasets Saved ===")

print("\nFull filtered dataset:")
print(f"Students: {full_check['user_id'].nunique():,}")
print(f"Interactions: {len(full_check):,}")

print("\nFiltered subsample:")
print(f"Students: {subsample_check['user_id'].nunique():,}")
print(f"Interactions: {len(subsample_check):,}")

print("\nAll files saved and verified!")
```
