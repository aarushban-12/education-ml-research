# Cleaning the ASSISTments 2017 dataset

```python
# Import the dataset and view the structure

import pandas as pd
import numpy as np

FILE_PATH = "assistments2017_raw.csv"

df = pd.read_csv(FILE_PATH, low_memory=False)

# Dataset dimensions
print("Shape:", df.shape)

# Log raw dataset
raw_rows = len(df)
raw_students = df["studentId"].nunique()

print("=== ASSISTments 2017: Raw Dataset ===")
print(f"Raw rows: {raw_rows:,}")
print(f"Raw students: {raw_students:,}")
print(f"Raw columns: {len(df.columns)}")

# Column names
print("\nColumns:")
print(df.columns.tolist())

# First 10 rows
display(df.head(10))
```

```python
# Rename columns
df = df.rename(columns={
    "studentId": "user_id",
    "problemId": "question_id",
    "skill": "skill_id",
    "startTime": "timestamp"
})

# Keep required columns
columns_to_keep = [
    "user_id",
    "question_id",
    "skill_id",
    "timestamp",
    "correct",
    "action_num",
    "original",
    "scaffold",
    "hintCount",
    "attemptCount",
    "MCAS"
]

df = df[columns_to_keep].copy()

# Verify
print("\n=== After Column Standardization ===")
print(f"Rows: {len(df):,}")
print(f"Students: {df['user_id'].nunique():,}")
print(f"Columns: {len(df.columns)}")

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

display(df.head(10))

print(df["timestamp"].head(10))
```

```python
# Required columns
required_cols = [
    "user_id",
    "question_id",
    "skill_id",
    "correct",
    "timestamp"
]

# Record initial counts
initial_rows = len(df)
initial_students = set(df["user_id"].dropna().unique())

# 1. Drop rows with missing required values
missing_mask = df[required_cols].isna().any(axis=1)
missing_removed = int(missing_mask.sum())

df = df.loc[~missing_mask].copy()

# 2. Drop rows where correct is not 0 or 1
valid_correct = df["correct"].isin([0, 1, "0", "1"])
invalid_correct_removed = int((~valid_correct).sum())

df = df.loc[valid_correct].copy()
df["correct"] = df["correct"].astype(int)

# 3. Drop rows where skill_id is "noskill"
noskill_mask = (
    df["skill_id"]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("noskill")
    .fillna(False)
)

noskill_removed = int(noskill_mask.sum())

df = df.loc[~noskill_mask].copy()

# Reset index
df = df.reset_index(drop=True)

# 4. Record students completely removed
remaining_students = set(df["user_id"].unique())
removed_students = initial_students - remaining_students

removed_students_df = pd.DataFrame({
    "user_id": sorted(removed_students)
})

# 5. Cleaning summary
print("=== ASSISTments 2017 Cleaning Summary ===")
print(f"Initial rows: {initial_rows:,}")
print(f"Missing required values removed: {missing_removed:,}")
print(f"Invalid correctness removed: {invalid_correct_removed:,}")
print(f"Noskill rows removed: {noskill_removed:,}")
print(f"Final rows: {len(df):,}")
print(f"Total rows removed: {initial_rows - len(df):,}")

print(f"\nInitial students: {len(initial_students):,}")
print(f"Remaining students: {len(remaining_students):,}")
print(f"Students completely removed: {len(removed_students):,}")

# 6. Verification
assert df[required_cols].notna().all().all()
assert df["correct"].isin([0, 1]).all()
assert not df["skill_id"].astype(str).str.strip().str.lower().eq("noskill").any()

print("\nAll cleaning checks passed!")

display(df.head(10))
```

```python
# Define event identifier
event_cols = ["user_id", "question_id", "timestamp"]

# Find events with conflicting correctness labels
conflict_counts = (
    df.groupby(event_cols)["correct"]
    .nunique()
)

conflicting_events = conflict_counts[
    conflict_counts > 1
].index

# Identify all rows belonging to conflicting events
conflict_mask = (
    pd.MultiIndex.from_frame(df[event_cols])
    .isin(conflicting_events)
)

# Save conflicting rows for documentation
df_conflicting = df.loc[conflict_mask].copy()

print("Conflicting events:", len(conflicting_events))
print("Rows involved:", len(df_conflicting))

display(df_conflicting.head(20))
```

```python
rows_before = len(df)

df = df.loc[~conflict_mask].copy().reset_index(drop=True)

print(f"Rows before: {rows_before:,}")
print(f"Rows removed: {rows_before - len(df):,}")
print(f"Rows remaining: {len(df):,}")

# Verify no conflicting events remain
assert (
    df.groupby(event_cols)["correct"].nunique() <= 1
).all()

print("All conflicting correctness events removed!")
```

```python
# Record counts before deduplication
rows_before = len(df)
students_before = df["user_id"].nunique()

# Count exact duplicates
duplicate_count = df.duplicated().sum()

# Remove exact duplicate rows
df = df.drop_duplicates().reset_index(drop=True)

# Cleaning summary
print("=== Duplicate Removal Summary ===")
print(f"Rows before: {rows_before:,}")
print(f"Exact duplicates removed: {duplicate_count:,}")
print(f"Rows remaining: {len(df):,}")
print(f"Students before: {students_before:,}")
print(f"Students remaining: {df['user_id'].nunique():,}")

# Verify
assert not df.duplicated().any()

print("\nAll exact duplicate rows removed!")
```

```python
# Preserve original row identity
df_check = df.copy().reset_index(drop=True)
df_check["_row_id"] = np.arange(len(df_check))

# Sort by timestamp, breaking ties with action_num
by_timestamp = (
    df_check.sort_values(
        ["user_id", "timestamp", "action_num", "_row_id"],
        kind="mergesort"
    )
    .reset_index(drop=True)
)

# Sort by action_num, breaking ties with timestamp
by_action = (
    df_check.sort_values(
        ["user_id", "action_num", "timestamp", "_row_id"],
        kind="mergesort"
    )
    .reset_index(drop=True)
)

# Compare the row order within each student
timestamp_order = (
    by_timestamp.groupby("user_id")["_row_id"]
    .apply(tuple)
)

action_order = (
    by_action.groupby("user_id")["_row_id"]
    .apply(tuple)
)

# Identify students with different orderings
different_students = (
    timestamp_order != action_order
)

n_different = int(different_students.sum())

print("=== Ordering Comparison ===")
print(f"Total students: {len(different_students):,}")
print(f"Students with identical ordering: {(~different_students).sum():,}")
print(f"Students with different ordering: {n_different:,}")
print(f"Percentage with differences: {different_students.mean() * 100:.2f}%")

# Show example students with differences
different_student_ids = different_students[
    different_students
].index.tolist()

for student_id in different_student_ids[:5]:
    print(f"\n=== Student {student_id} ===")

    print("\nSorted by timestamp:")
    display(
        by_timestamp.loc[
            by_timestamp["user_id"] == student_id,
            ["question_id", "timestamp", "action_num", "correct"]
        ].head(15)
    )

    print("Sorted by action_num:")
    display(
        by_action.loc[
            by_action["user_id"] == student_id,
            ["question_id", "timestamp", "action_num", "correct"]
        ].head(15)
    )
```

```python
# Sort chronologically by student, timestamp, and action_num
df = (
    df.sort_values(
        ["user_id", "timestamp", "action_num"],
        kind="mergesort"
    )
    .reset_index(drop=True)
)

# Verify chronological ordering
assert df.groupby("user_id")["timestamp"].apply(
    lambda x: x.is_monotonic_increasing
).all()

print("Chronological sorting complete!")
print(f"Students: {df['user_id'].nunique():,}")
print(f"Rows: {len(df):,}")

display(df.head(10))
```

```python
# 1. Extract student-level MCAS data
df_mcas = df[["user_id", "MCAS"]].copy()

# Remove missing MCAS values
df_mcas = df_mcas.dropna(subset=["MCAS"])

# Check whether students have multiple distinct MCAS values
mcas_counts = df_mcas.groupby("user_id")["MCAS"].nunique()

print("=== MCAS Consistency Check ===")
print(f"Students with MCAS data: {df_mcas['user_id'].nunique():,}")
print(f"Students with multiple MCAS values: {(mcas_counts > 1).sum():,}")

# Display any inconsistencies
if (mcas_counts > 1).any():
    inconsistent_ids = mcas_counts[mcas_counts > 1].index
    display(df_mcas[df_mcas["user_id"].isin(inconsistent_ids)].head(20))
```

```python
# 2. Create one MCAS record per student
df_mcas = (
    df_mcas.drop_duplicates(subset=["user_id"])
           .sort_values("user_id")
           .reset_index(drop=True)
)

# 3. Save MCAS separately
df_mcas.to_parquet("assistments2017_student_mcas.parquet", index=False)

# 4. Remove MCAS from main interaction dataset
df = df.drop(columns=["MCAS"])

# 5. Verify
assert df_mcas["user_id"].is_unique
assert "MCAS" not in df.columns

print("MCAS dataset saved successfully!")
print(f"Students with MCAS: {len(df_mcas):,}")
print(f"Main dataset rows: {len(df):,}")
```

```python
# Calculates each student's early window as the larger of 40% of their 
# interactions or the number needed to reach 20 unique questions. 
# It also counts remaining interactions to determine eligibility for later evaluation.
def early_window_lengths(
    df,
    student_col="user_id",
    question_col="question_id",
    position_col="action_num",
    early_fraction=0.40,
    min_unique_questions=20
):
    results = []

    for student_id, group in df.groupby(student_col, sort=False):
        group = group.sort_values(position_col)

        n_rows = len(group)
        n_unique_questions = group[question_col].nunique()

        # Minimum number of rows covering 40% of interactions
        fraction_length = int(np.ceil(early_fraction * n_rows))

        # Find the row where the 20th distinct question appears
        first_encounters = ~group[question_col].duplicated()

        if n_unique_questions >= min_unique_questions:
            twentieth_index = np.flatnonzero(
                first_encounters.to_numpy()
            )[min_unique_questions - 1]

            rows_to_twentieth = int(twentieth_index + 1)

            early_length = max(
                fraction_length,
                rows_to_twentieth
            )
        else:
            rows_to_twentieth = pd.NA
            early_length = pd.NA

        remaining_rows = (
            n_rows - early_length
            if pd.notna(early_length)
            else pd.NA
        )

        results.append({
            student_col: student_id,
            "total_rows": n_rows,
            "unique_questions": n_unique_questions,
            "fraction_length": fraction_length,
            "rows_to_20th_question": rows_to_twentieth,
            "early_window_length": early_length,
            "remaining_rows": remaining_rows
        })

    return pd.DataFrame(results)
```

```python
# Calculate early-window lengths
early_windows = early_window_lengths(df)

# Assign exclusion reasons
early_windows["exclusion_reason"] = np.select(
    [
        early_windows["unique_questions"] < 20,
        early_windows["remaining_rows"].fillna(-1) < 5
    ],
    [
        "fewer than 20 different questions",
        "fewer than 5 rows left"
    ],
    default=""
)

early_windows["eligible"] = (
    early_windows["exclusion_reason"] == ""
)

# Keep eligible student summaries
kept_students_df = early_windows[
    early_windows["eligible"]
].copy()

# Keep excluded student summaries and reasons
excluded_students_df = early_windows[
    ~early_windows["eligible"]
].copy()

# Filter the canonical interaction dataset
df_filtered = df[
    df["user_id"].isin(
        kept_students_df["user_id"]
    )
].copy().reset_index(drop=True)

print("Students before:", df["user_id"].nunique())
print("Students retained:", len(kept_students_df))
print("Students excluded:", len(excluded_students_df))

print("\nRows before:", len(df))
print("Rows retained:", len(df_filtered))
print("Rows excluded:", len(df) - len(df_filtered))

print("\nExclusion reasons:")
print(excluded_students_df["exclusion_reason"].value_counts())

display(excluded_students_df.head(10))
```
