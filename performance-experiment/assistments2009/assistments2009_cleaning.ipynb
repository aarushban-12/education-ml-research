# Cleaning the ASSISTments 2009 dataset

```python
# Import the dataset and view the structure

import pandas as pd
import numpy as np

FILE_PATH = "assistments2009_raw.csv"

df = pd.read_csv(FILE_PATH, encoding='latin1')

# Dataset dimensions
print("Shape:", df.shape)

# Column names
print("\nColumns:")
print(df.columns.tolist())

# First 10 rows
display(df.head(10))
```

```python
# Drop columns that are not needed for analysis
columns_to_keep = [
    "user_id",
    "problem_id",
    "skill_id",
    "correct",
    "order_id",
    "original",
    "assistment_id",
    "attempt_count",
    "hint_count"
]

df = df[columns_to_keep].copy()

print("Shape after dropping columns:", df.shape)
display(df.head(10))
```

```python
# Make sure that the correct column only contains 0 and 1 values
print("Unique values:", df["correct"].unique())

print(df["correct"].value_counts(dropna=False))
print("\nOnly contains 0 and 1:", df["correct"].isin([0, 1]).all())


required_columns = [
    "user_id",
    "problem_id",
    "skill_id",
    "correct",
    "order_id"
]

# Count rows before cleaning
rows_before = len(df)

# Drop rows with missing required values
df = df.dropna(subset=required_columns).copy()

rows_after = len(df)

print(f"Rows before: {rows_before:,}")
print(f"Rows after: {rows_after:,}")
print(f"Rows dropped: {rows_before - rows_after:,}")

# Verify no missing values remain
print("\nMissing values in required columns:")
print(df[required_columns].isna().sum())
```

```python
# Count rows before removing duplicates
rows_before = len(df)

# Check number of duplicate rows
print(f"Duplicate rows found: {df.duplicated().sum():,}")

# Remove duplicate rows
df = df.drop_duplicates().copy()

# Reset index
df.reset_index(drop=True, inplace=True)

print(f"Rows before: {rows_before:,}")
print(f"Rows after: {len(df):,}")
print(f"Rows removed: {rows_before - len(df):,}")
print(f"Remaining duplicates: {df.duplicated().sum():,}")
```

```python
# Check how many problems and rows are associated with multiple skills
skills_per_problem = df.groupby("problem_id")["skill_id"].nunique()

print("Problems associated with multiple skills:",
      (skills_per_problem > 1).sum())

print("Rows belonging to those problems:",
      df["problem_id"].isin(
          skills_per_problem[skills_per_problem > 1].index
      ).sum())

display(skills_per_problem[skills_per_problem > 1].head(20))
```

```python
# Group all unique skills associated with each problem
skill_combinations = (
    df.groupby("problem_id")["skill_id"]
    .apply(lambda x: tuple(sorted(set(x))))
)

# Count each unique combination
combination_counts = skill_combinations.value_counts()

print("Total unique skill combinations:", len(combination_counts))
print("Combinations containing multiple skills:",
      sum(len(combo) > 1 for combo in combination_counts.index))

# Display the most common combinations
display(combination_counts.head(20))
```

```python
# Count only the number of unique combinations that contain multiple skills
multi_skill_combinations = combination_counts[
    combination_counts.index.map(lambda combo: len(combo) > 1)
]

print("Unique multi-skill combinations:", len(multi_skill_combinations))

display(multi_skill_combinations.head(30))
```

```python
# Count distinct skills associated with each problem
skills_per_problem = (
    df.groupby("problem_id")["skill_id"]
    .nunique()
)

# Identify problems with multiple skills
multi_skill_problem_ids = skills_per_problem[
    skills_per_problem > 1
].index

print(f"Problems with multiple skills: {len(multi_skill_problem_ids):,}")

# Select the first 10 multi-skill problems
sample_problem_ids = multi_skill_problem_ids[:10]

# Display their interactions
sample = df[
    df["problem_id"].isin(sample_problem_ids)
].sort_values(["problem_id", "user_id", "order_id"])

display(
    sample[
        ["problem_id", "user_id", "order_id",
         "skill_id", "correct", "attempt_count"]
    ].head(50)
)
```

```python
# Check for interactions with multiple skills and conflicting correctness values
interaction_cols = ["user_id", "order_id", "problem_id"]

interaction_summary = (
    df.groupby(interaction_cols)
    .agg(
        rows=("skill_id", "size"),
        unique_skills=("skill_id", "nunique"),
        unique_correct=("correct", "nunique")
    )
    .reset_index()
)

multi_skill_interactions = interaction_summary[
    interaction_summary["unique_skills"] > 1
]

print("Total unique interactions:", len(interaction_summary))
print("Interactions with multiple skills:", len(multi_skill_interactions))

print(
    "Percentage with multiple skills:",
    round(100 * len(multi_skill_interactions) / len(interaction_summary), 2)
)

print(
    "Multi-skill interactions with conflicting correctness:",
    (multi_skill_interactions["unique_correct"] > 1).sum()
)

display(multi_skill_interactions.head(20))
```

```python
# Check that repeated rows agree on response and metadata
metadata_cols = [
    "correct",
    "original",
    "assistment_id",
    "attempt_count",
    "hint_count"
]

for col in metadata_cols:
    conflicts = (
        df.groupby(interaction_cols)[col]
        .nunique(dropna=False)
        .gt(1)
        .sum()
    )
    print(f"{col} conflicts: {conflicts:,}")

# Stop if metadata differs within the same interaction
assert all(
    df.groupby(interaction_cols)[col]
    .nunique(dropna=False)
    .le(1)
    .all()
    for col in metadata_cols
), "Conflicting metadata found. Inspect before merging."

# Combine skills for each interaction
skill_groups = (
    df.groupby(interaction_cols)["skill_id"]
    .apply(lambda x: tuple(sorted(set(int(v) for v in x))))
    .reset_index(name="skill_combination")
)

# Keep one response row per interaction
df_interactions = (
    df.drop(columns=["skill_id"])
    .drop_duplicates(subset=interaction_cols)
    .merge(
        skill_groups,
        on=interaction_cols,
        how="left",
        validate="one_to_one"
    )
)

# Create stable, reproducible IDs for each unique skill combination
unique_combinations = sorted(
    set(df_interactions["skill_combination"])
)

combination_mapping = {
    combo: i + 1
    for i, combo in enumerate(unique_combinations)
}

# Use a list comprehension instead of pandas .map()
df_interactions["skill_combination_id"] = [
    combination_mapping[combo]
    for combo in df_interactions["skill_combination"]
]

# Sort chronologically
df_interactions = (
    df_interactions
    .sort_values(["user_id", "order_id", "problem_id"])
    .reset_index(drop=True)
)

# Verify
print("Total interactions:", len(df_interactions))
print("Unique skill combinations:", len(unique_combinations))
print("Unique combination IDs:", df_interactions["skill_combination_id"].nunique())

display(
    df_interactions[
        ["user_id", "problem_id", "skill_combination", "skill_combination_id"]
    ].head(20)
)
```

```python
def canonical_sort(df, sort_columns):
    # Stable sort
    df = (
        df.sort_values(
            by=sort_columns,
            ascending=True,
            kind="mergesort"
        )
        .reset_index(drop=True)
        .copy()
    )

    # Assign sequential position within each student
    df["position"] = (
        df.groupby("user_id").cumcount()
    )

    return df


# Apply canonical sorting
df_interactions = canonical_sort(
    df_interactions,
    ["user_id", "order_id", "problem_id"]
)

display(
    df_interactions.head(30)
)
```

```python
# 1. Re-sorting produces exactly the same dataset
resorted = canonical_sort(
    df_interactions,
    ["user_id", "order_id", "problem_id"]
)

assert df_interactions.equals(resorted), (
    "Re-sorting changed the dataset!"
)

# 2. Position starts at 0 for every student
assert (
    df_interactions.groupby("user_id")["position"].min() == 0
).all()

# 3. Position ends at n-1 for every student
student_counts = df_interactions.groupby("user_id").size()

student_max_positions = (
    df_interactions.groupby("user_id")["position"].max()
)

assert (
    student_max_positions == student_counts - 1
).all()

# 4. No duplicate positions within a student
assert not df_interactions.duplicated(
    subset=["user_id", "position"]
).any()

# 5. Confirm sorted keys
assert df_interactions[
    ["user_id", "order_id", "problem_id"]
].equals(
    df_interactions[
        ["user_id", "order_id", "problem_id"]
    ].sort_values(
        ["user_id", "order_id", "problem_id"],
        kind="mergesort"
    ).reset_index(drop=True)
)

print("All canonical sorting checks passed!")
```

```python
# Calculates each student's early window as the larger of 40% of their 
# interactions or the number needed to reach 20 unique questions. 
# It also counts remaining interactions to determine eligibility for later evaluation.
def early_window_lengths(
    df,
    student_col="user_id",
    question_col="problem_id",
    position_col="position",
    early_fraction=0.40,
    min_unique_questions=20
):
    results = []

    for student_id, group in df.groupby(student_col, sort=False):
        group = group.sort_values(position_col)

        n_rows = len(group)
        n_unique_questions = group[question_col].nunique()

        # Minimum number of rows covering 40% of interactions
        fraction_length = int(np.ceil(early_fraction * n_rows))

        # Find the row where the 20th distinct question appears
        first_encounters = ~group[question_col].duplicated()

        if n_unique_questions >= min_unique_questions:
            twentieth_index = np.flatnonzero(
                first_encounters.to_numpy()
            )[min_unique_questions - 1]

            rows_to_twentieth = int(twentieth_index + 1)

            early_length = max(
                fraction_length,
                rows_to_twentieth
            )
        else:
            rows_to_twentieth = pd.NA
            early_length = pd.NA

        remaining_rows = (
            n_rows - early_length
            if pd.notna(early_length)
            else pd.NA
        )

        results.append({
            student_col: student_id,
            "total_rows": n_rows,
            "unique_questions": n_unique_questions,
            "fraction_length": fraction_length,
            "rows_to_20th_question": rows_to_twentieth,
            "early_window_length": early_length,
            "remaining_rows": remaining_rows
        })

    return pd.DataFrame(results)
```

```python
# Calculate early-window lengths
early_windows = early_window_lengths(df_interactions)

# Assign exclusion reasons
early_windows["exclusion_reason"] = np.select(
    [
        early_windows["unique_questions"] < 20,
        early_windows["remaining_rows"].fillna(-1) < 5
    ],
    [
        "fewer than 20 different questions",
        "fewer than 5 rows left"
    ],
    default=""
)

early_windows["eligible"] = (
    early_windows["exclusion_reason"] == ""
)

# Keep eligible student summaries
kept_students_df = early_windows[
    early_windows["eligible"]
].copy()

# Keep excluded student summaries and reasons
excluded_students_df = early_windows[
    ~early_windows["eligible"]
].copy()

# Filter the canonical interaction dataset
df_filtered = df_interactions[
    df_interactions["user_id"].isin(
        kept_students_df["user_id"]
    )
].copy().reset_index(drop=True)

print("Students before:", df_interactions["user_id"].nunique())
print("Students retained:", len(kept_students_df))
print("Students excluded:", len(excluded_students_df))

print("\nRows before:", len(df_interactions))
print("Rows retained:", len(df_filtered))
print("Rows excluded:", len(df_interactions) - len(df_filtered))

print("\nExclusion reasons:")
print(excluded_students_df["exclusion_reason"].value_counts())

display(excluded_students_df.head(10))
```

```python
# Verification of the student filtering
# Every retained student has at least 20 distinct questions
assert (kept_students_df["unique_questions"] >= 20).all()

# Every retained student has at least 5 rows after early window
assert (kept_students_df["remaining_rows"] >= 5).all()

# Every excluded student fails at least one condition
assert (
    (excluded_students_df["unique_questions"] < 20)
    |
    (excluded_students_df["remaining_rows"].fillna(-1) < 5)
).all()

# Student counts match
assert (
    len(kept_students_df) + len(excluded_students_df)
    == df_interactions["user_id"].nunique()
)

# Row counts match
assert (
    kept_students_df["total_rows"].sum()
    == len(df_filtered)
)

assert (
    kept_students_df["total_rows"].sum()
    + excluded_students_df["total_rows"].sum()
    == len(df_interactions)
)

# Ensure original positions were preserved
assert not df_filtered.duplicated(
    ["user_id", "position"]
).any()

print("All filtering checks passed!")
```

```python
def inspect_student(student_id):
    group = (
        df_interactions[
            df_interactions["user_id"] == student_id
        ]
        .sort_values("position")
        .copy()
    )

    summary = early_windows.loc[
        early_windows["user_id"] == student_id
    ].iloc[0]

    group["first_encounter"] = (
        ~group["problem_id"].duplicated()
    )

    group["cumulative_unique_questions"] = (
        group["first_encounter"].cumsum()
    )

    early_length = summary["early_window_length"]

    if pd.notna(early_length):
        group["in_early_window"] = (
            np.arange(len(group)) < int(early_length)
        )
    else:
        group["in_early_window"] = pd.NA

    print(f"\nStudent: {student_id}")
    print(f"Total rows: {summary['total_rows']}")
    print(f"Unique questions: {summary['unique_questions']}")
    print(f"40% length: {summary['fraction_length']}")
    print(f"Rows to 20th question: {summary['rows_to_20th_question']}")
    print(f"Early window length: {early_length}")
    print(f"Remaining rows: {summary['remaining_rows']}")
    print(f"Eligible: {summary['eligible']}")
    print(f"Reason: {summary['exclusion_reason'] or 'Retained'}")

    display(
        group[
            [
                "position",
                "order_id",
                "problem_id",
                "first_encounter",
                "cumulative_unique_questions",
                "in_early_window"
            ]
        ]
    )

# One retained student
if not kept_students_df.empty:
    inspect_student(kept_students_df.iloc[0]["user_id"])

# One excluded student for each reason
for reason in [
    "fewer than 20 different questions",
    "fewer than 5 rows left"
]:
    candidates = excluded_students_df[
        excluded_students_df["exclusion_reason"] == reason
    ]

    if not candidates.empty:
        inspect_student(candidates.iloc[0]["user_id"])
    else:
        print(f"\nNo students excluded for: {reason}")
```

```python
# Save eligible interaction data
df_filtered.to_parquet(
    "assistments2009_filtered.parquet",
    index=False
)

# Save complete student-level eligibility audit
early_windows.to_csv(
    "assistments2009_eligibility_audit.csv",
    index=False
)

# Save excluded students separately
excluded_students_df.to_csv(
    "assistments2009_excluded_students.csv",
    index=False
)
```
